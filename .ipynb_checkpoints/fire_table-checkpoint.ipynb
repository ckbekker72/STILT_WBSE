```python
import os
import glob
import datetime
import geopandas as gpd
import numpy as np
import pandas as pd
import datetime
from osgeo import gdal
from datetime import datetime, date, timedelta
```

```python
def total_ems(grd_path):
    dataset = gdal.Open(grd_path)
    band_num = 12 #The band number for PM2_5
    band = dataset.GetRasterBand(band_num)
    data = band.ReadAsArray()
    data = np.where(np.isnan(data), 0, data) #Replace nan values with 0 
    total = np.sum(data) #Total emissions
    return total
```

```python
def burn_severity(raster_path):
  ds = gdal.Open(raster_path)
  myarray = np.array(ds.GetRasterBand(1).ReadAsArray()) #Raster data into an array
  non_processing_area = (myarray==0).sum()
  unburned = (myarray==1).sum()
  low = (myarray==2).sum()
  medium = (myarray==3).sum()
  high = (myarray==4).sum()
  grass = (myarray==5).sum()
  total_area = non_processing_area + unburned + low + medium + high + grass
  low = (low / total_area) * 100
  medium = (medium / total_area) * 100
  high = (high / total_area) * 100
  return low, medium, high
```

```python
def central(data):
    data["id"] = 1
    dissolved = data.dissolve(by = 'id', aggfunc = 'sum')
    centroid = dissolved.to_crs('+proj=cea').centroid.to_crs("EPSG:4326")
    return centroid 
```

```python
def starting_day(data):
  year = "2018"
  day_num = data['dob'].min()

  strt_date = date(int(year), 1, 1)

  res_date = strt_date + timedelta(days=int(day_num) - 1)
  res = res_date.strftime("%m-%d-%Y")

  return res
```

```python
def ending_day(data): 
  year = "2018"
  day_num = data['dob'].max()

  strt_date = date(int(year), 1, 1)

  res_date = strt_date + timedelta(days=int(day_num) - 1)
  res = res_date.strftime("%m-%d-%Y")

  return res
```

```python
def peak(data):
  data = data[["dob", "PM2_5"]]
  agg_functions = {'PM2_5': 'sum'}

  #create new DataFrame by combining rows with same id values
  df_new = data.groupby(data['dob'], as_index=False).aggregate(agg_functions).reindex(columns=data.columns)
  data = df_new.fillna(0)
  
  peak_day = data.loc[data.PM2_5 == data.PM2_5.max(), 'dob'].values[0]

  return peak_day
```

```python
fire = []
central_lat = []
central_lon = []
peak_emissions = []
total_emissions = []
low = []
medium = []
high = []
first_day = []
last_day = []

shp_files = sorted(glob.glob("WBSE_emissions/*.shp"),  key=os.path.getmtime)
for file in shp_files:
    files = glob.glob('output/*')
    # Read Polygon Shapefile
    vector = gpd.read_file(file)
    vector = vector[vector['dob'] != -9999.0]
    # Reprojecting to 4326 coordinate system
    vector = vector.to_crs('EPSG:4326')
    name = file[15:36]

    #raster path
    raster_path = "WBSE_burnsev_2018/" + name + "_burn_severity_CA.tif"
    #grd path
    grd_path = "WBSE_emis_2018/" + name + "_stack.grd" 
    grd_path = "WBSE_emis_2018/" + name + "_stack.grd"

    fire.append(name)
    lon = central(vector).x.iat[0]
    lat = central(vector).y.iat[0]
    central_lat.append(lat)
    central_lon.append(lon)
    first_day.append(starting_day(vector))
    last_day.append(ending_day(vector))
    peak_emissions.append(peak(vector))
    total_emissions.append(total_ems(grd_path))
    l, m, h = burn_severity(raster_path)
    low.append(l)
    medium.append(m)
    high.append(h)
```

```python
dict = {'fire': fire, 'longitude': central_lon, 'latitude': central_lat, 'First Day': first_day,
        'Last Day': last_day, 'Peak Emissions': peak_emissions, 'Total Emissions': total_emissions,
       'Low Severity (%)': low, 'Medium Severity (%)': medium, 'High Severity (%)': high}  
df = pd.DataFrame(dict) 
```
